# Functions

```python
#Creating the function
def add_one(number):
  return number + 1
```

```python
#Using the function or calling the function
add_one(3)
```

```python
def bp_check(systolic_bp):
  if systolic_bp > 120:
    bp_result = "High"
  else:
    bp_result = "Normal"
  return bp_result
```

```python
bp_check(150)
```

# Loops

```python
bp_list = [120, 135, 170, 110, 100]
```

```python
for bp in bp_list:
  print(bp_check(bp))
```

```python
def bmi_check(bmi):
  if bmi < 18.5:
    bmi_result = "Underweight"
  elif bmi < 25:
    bmi_result = "Normal"
  elif bmi < 30:
    bmi_result = "Overweight"
  else:
    bmi_result = "Obese"
  return bmi_result
```

#Loops allow us to change values to an easy to read label.

```python
bmi_check(24.9)
```

```python
#Group 2
def calculate_bmi(height, weight):
  bmi = weight/height ** 2
  return bmi
```

```python
calculate_bmi(1.2,57)
```

```python
calculate_bmi(height = 1.2, weight = 57)
```
